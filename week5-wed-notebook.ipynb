## 2. Cosine Similarity Whiteboard Activity

This activity was completed by hand, here is how to do it via Python code.

### Inputs for the problem

````python
documents = [
    "Max likes Sam.",    # D1
    "Sam likes Max.",    # D2
    "Max and Sam.",      # D3
    "Winter likes fall." # D4
]

query = "Sam likes winter"
````

### Step 1: Create vocabulary of all terms

First I will concatenate all strings into one.

````python
allText = " ".join(documents + [query])
print(allText)
````

I don't want to have punctuation. There are several ways to do that. Let's see some:

#### Method 1: Iterate over the string, leave out punctuations

````python
import string
print(string.punctuation)
````

````python
allText_noPunct = "".join([ch for ch in allText if not ch in string.punctuation])
print(allText_noPunct)
````

### Method 2: Use built-in methods

Pythyn's built-in `str` class has many useful methods for translating a set of characters to another one. Let's see the method in action:

````python
allText.translate(str.maketrans("", "", string.punctuation))
````

What's going on?

````python
print(str.maketrans("", "", string.punctuation))
````

We can see that all ASCII Characters from 0-32 and from 65 to 90 are missing, because those characters will be deleted.

Here is what is happening:

```
str.maketrans(from_str, to_str, delete_chars)
#              ^          ^        ^
#           1st arg    2nd arg  3rd arg
```

* 1st argument (""): The characters you want to replace. (Empty string = replace nothing)

* 2nd argument (""): The characters to replace them with. (Empty string = no replacement mapping)

* 3rd argument (string.punctuation): The characters you want to delete entirely.

### Our vocabulary

Here is a Python one-liner that takes our variable `allText_noPunct` and find the unique words in it.

````python
words = set(allText_noPunct.lower().split())
words
````

````python
voc = sorted(words)
voc
````

### YOUR TURN

Write code to implement the other steps of the process shown in Slide 7.

**Show your process before writing any functions.**

**Use markdown to label the three remaining steps.**

````python
def toVector(text):
    text = text.lower().replace(".", "")
    words = text.split()
    vector = []
    for term in voc:
        vector.append(words.count(term))
    return vector

docVectors = []
for d in documents:
    docVectors.append(toVector(d))

queryVector = toVector(query)

print(docVectors)
print(queryVector)
````

````python
import math

def cosine(a, b):
    dot = 0
    for i in range(len(a)):
        dot = dot + a[i] * b[i]
    lenA = math.sqrt(sum(x * x for x in a))
    lenB = math.sqrt(sum(x * x for x in b))
    return dot / (lenA * lenB)

scores = []
for v in docVectors:
    scores.append(cosine(v, queryVector))

print(scores)
````

````python
pairs = list(zip(documents, scores))
pairs.sort(key=lambda p: p[1], reverse=True)

for doc, score in pairs:
    print(doc, round(score, 3))
````

````python
headers={'User-Agent': 'CS315-Student/1.0 (contact: your_email@wellesley.edu)'}
def check_title(title):
    url= "https://en.wikipedia.org/w/api.php"
    params= {"action": "query", "titles": title, "redirects": 1, "format": "json"}
    r= requests.get(url, params=params, headers=headers).json()
    page= list(r["query"]["pages"].values())[0]
    if "missing" in page:
        return "MISSING"
    return page["title"].replace(" ", "_")
for show, title in wiki_titles.items():
    print(f"{show:30} -> {title:35} -> {check_title(title)}")
````

````python
target_titles = set(wiki_titles.values())
target_titles
````

Downloading one day of data
- **Cache:** if the day is already saved as a CSV, load it instead of re-downloading.
- **Stream:** read the huge file one line at a time to save memory.
- **Filter:** keep only English Wikipedia rows for our 10 shows.

````python
cols= ['country', 'country_code', 'project', 'page_id', 'page_title', 'item_id', 'pageviews']
import os
os.makedirs("daily", exist_ok=True)
def get_day(date_str):
    cache= f"daily/{date_str}.csv"
    if os.path.exists(cache):
        return pd.read_csv(cache)
    base= "https://analytics.wikimedia.org/published/datasets/country_project_page/"
    url= base + date_str + ".tsv"
    keep=[]
    with requests.get(url, headers=headers, stream=True) as response:
        for line in response.iter_lines(decode_unicode=True):
            parts= line.split('\t')
            if len(parts)==7 and parts[2]=='en.wikipedia' and parts[4] in target_titles:
                keep.append(parts)
    df= pd.DataFrame(keep, columns=cols)
    df['pageviews']= df['pageviews'].astype(int)
    df['date']= date_str
    df.to_csv(cache, index=False)
    return df
````

This takes a long time to run

````python
from datetime import date, timedelta
frames= []
day= date(2026, 9, 1)
while day<=date(2026, 9, 30):
    d= day.strftime("%Y-%m-%d")
    print("getting", d)
    frames.append(get_day(d))
    day = day + timedelta(days=1)
sept = pd.concat(frames, ignore_index=True)
print(sept.shape)
sept.head()
````

````python
title_to_show = {v: k for k, v in wiki_titles.items()}
sept['show'] = sept['page_title'].map(title_to_show)
sept.to_csv("kdrama_pageviews_sept2026.csv", index=False)
````
